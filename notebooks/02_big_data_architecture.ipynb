```mermaid
flowchart TD
    A[Raw Data\nKaggle CSV] 
    --> B[Data Ingestion\nPython + pandas]
    --> C[Data Cleaning & Validation]
    --> D[Feature Engineering]
    --> E[Data Storage\nSQLite + CSV + Parquet]
    --> F[Spark Processing\nDistributed Computing]
    --> G[Analytics & SQL Queries]
    --> H[Tableau Dashboard\nBusiness Insights]
    --> I[Deployment\nDocker + Cloud Azure]

## Erklärung der Pipeline-Schritte

**1. Raw Data**  
→ Die ursprüngliche Kaggle-CSV-Datei (24.767 Wohnungen)

**2. Data Ingestion**  
→ Laden mit Python + pandas

**3. Data Cleaning & Feature Engineering**  
→ Bereinigung, Behandlung fehlender Werte, Erstellung neuer Features (price_per_room, building_age_group etc.)

**4. Data Storage**  
→ Gespeichert als CSV, Parquet und in SQLite-Datenbank

**5. Spark Processing (Big Data Layer)**  
→ Für zukünftige Skalierung auf große Datenmengen (verteilte Verarbeitung)

**6. Analytics & SQL**  
→ Komplexe Abfragen für Business Insights

**7. Visualization**  
→ Tableau Dashboard für End-User

**8. Deployment**  
→ Docker + Cloud (Azure) für produktiven Einsatz

## 4.2 Hadoop & Spark Konzept

### Warum Big Data Technologien?

Auch wenn unser aktuelles Dataset mit 24.767 Zeilen noch gut mit Pandas zu verarbeiten ist, muss ein modernes Data Engineering Projekt auf **Skalierbarkeit** vorbereitet sein. Bei Millionen von Datensätzen oder Echtzeit-Updates stoßen Pandas und SQLite an ihre Grenzen.

### Hadoop – Das Fundament

- **HDFS (Hadoop Distributed File System)**: Verteilt große Datenmengen auf mehrere Server
- **Hohe Fehlertoleranz**: Daten bleiben erhalten, auch wenn einzelne Knoten ausfallen
- **Skalierbar**: Von wenigen GB bis Petabytes

### Spark – Die moderne Verarbeitungs-Engine

- **In-Memory Processing**: Bis zu 100x schneller als klassisches Hadoop MapReduce
- **PySpark**: Python-API, die wir später nutzen können
- **Unterstützt**: Batch, Streaming, Machine Learning, SQL

**Unser geplanter Einsatz:**
- Spark für Feature Engineering bei wachsenden Datenmengen
- Spark SQL für komplexe Analysen
- Parquet-Format als Speicherformat (sehr effizient)

## 4.3 Cloud Konzept

### Warum Cloud?

Lokale Systeme stoßen bei wachsenden Datenmengen und häufigen Updates schnell an Grenzen. Die Cloud bietet **Skalierbarkeit**, **Verfügbarkeit** und **Kosteneffizienz**.

### Geplante Cloud-Architektur (Azure-fokussiert)

**Speicherung**
- **Azure Data Lake Storage Gen2** → für Rohdaten und Parquet-Dateien
- **Azure SQL Database** oder **Azure Synapse** → für strukturierte Analysen

**Verarbeitung**
- **Azure Databricks** → managed Spark-Umgebung (PySpark)
- **Azure Data Factory** → für Orchestrierung der Pipeline

**Deployment & Automation**
- **Docker** → Containerisierung der Anwendung
- **Azure Container Apps** oder **Azure Kubernetes Service (AKS)** → für produktiven Betrieb
- **GitHub Actions** → CI/CD Pipeline

**Vorteile dieser Architektur**
- Automatische Skalierung bei steigender Datenmenge
- Pay-as-you-go (nur bezahlen, was genutzt wird)
- Hohe Verfügbarkeit und Backup-Möglichkeiten
- Einfache Integration mit Tableau (über Direct Query oder Extracts)
