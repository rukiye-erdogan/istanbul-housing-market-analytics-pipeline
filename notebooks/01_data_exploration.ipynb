```python
import pandas as pd
import numpy as np
import os

print("Bibliotheken erfolgreich importiert!")
```

```python
# Pfad zur Datei
file_path = "../01_data/istanbul_apartment_prices_2026.csv"

# Prüfen, ob die Datei existiert
if os.path.exists(file_path):
    print("Datei gefunden!")
else:
    print("Datei nicht gefunden. Bitte Pfad überprüfen.")

# Datei laden
df = pd.read_csv(file_path)

print(f"Datei erfolgreich geladen! Shape: {df.shape}")
```

```python
# Grundinformationen
df.info()
```

```python
# Statistische Zusammenfassung
df.describe()
```

```python
# Alle Spaltennamen anzeigen
print("Spalten im Dataset:")
print(df.columns.tolist())

# Fehlende Werte pro Spalte (in Prozent)
missing = df.isnull().sum().sort_values(ascending=False)
missing_percent = (missing / len(df) * 100).round(2)

print("\nFehlende Werte (sortiert):")
print(pd.concat([missing, missing_percent], axis=1, keys=['Fehlend', '%']))
```

```python
# Beispiele für wichtige Spalten
print(df['district'].value_counts().head(10))
print("\n")
print(df['price'].describe())
```

```python
# Sicherheitskopie des Original-DataFrames erstellen
df_original = df.copy()

print("Original-DataFrame gesichert!")
print(f"Shape vor Cleaning: {df.shape}")
```

```python
# 1. Fehlende Werte behandeln

# Spalten mit sehr vielen fehlenden Werten (über 60%) erstmal mit "Unbekannt" füllen
df['complex_name'] = df['complex_name'].fillna('Unbekannt')
df['floor_category'] = df['floor_category'].fillna('Unbekannt')
df['maintenance_fee'] = df['maintenance_fee'].fillna(0)

# Bei building_type und fuel_type ebenfalls "Unbekannt"
df['building_type'] = df['building_type'].fillna('Unbekannt')
df['fuel_type'] = df['fuel_type'].fillna('Unbekannt')
df['building_condition'] = df['building_condition'].fillna('Unbekannt')

print("Fehlende Werte in wichtigen Spalten behandelt!")
print(f"Neue Shape: {df.shape}")
```

```python
# 2. Duplikate prüfen und entfernen (falls vorhanden)

# Anzahl Duplikate prüfen
duplicates = df.duplicated().sum()
print(f"Anzahl doppelter Zeilen: {duplicates}")

# Falls Duplikate vorhanden → entfernen
if duplicates > 0:
    df = df.drop_duplicates()
    print("Duplikate entfernt!")
else:
    print("Keine Duplikate gefunden.")
    
print(f"Shape nach Duplikat-Check: {df.shape}")
```

```python
# 3. Datentypen optimieren

print("Aktuelle Datentypen:")
print(df.dtypes)

# Kategorische Spalten optimieren (spart Speicher und ist gut für Analyse)
categorical_cols = ['district', 'neighborhood', 'building_type', 'building_condition', 
                    'heating_type', 'fuel_type', 'usage_status', 'credit_eligible', 
                    'deed_status', 'exchange', 'floor_category']

for col in categorical_cols:
    if col in df.columns:
        df[col] = df[col].astype('category')

print("\n Kategorische Spalten optimiert!")
```

```python
# 4. Restliche kleine fehlende Werte behandeln

# Numerische Spalten mit Median füllen
df['floor'] = df['floor'].fillna(df['floor'].median())
df['bathroom_count'] = df['bathroom_count'].fillna(df['bathroom_count'].median())

# Bei orientation "Unbekannt" falls noch fehlend
df['orientation'] = df['orientation'].fillna('Unbekannt')

# Letzte Überprüfung
print("Finales Data Cleaning abgeschlossen!")
print(f"Endgültige Shape: {df.shape}")

# Übersicht über verbleibende fehlende Werte
remaining_missing = df.isnull().sum().sum()
print(f"Verbleibende fehlende Werte insgesamt: {remaining_missing}")
```

```python
# Saubere Version speichern
df.to_csv('../01_data/istanbul_apartment_prices_cleaned.csv', index=False)

print("Cleaned Dataset erfolgreich gespeichert!")
print("Datei: ../01_data/istanbul_apartment_prices_cleaned.csv")
```

```python
# 1. Preis pro Zimmer berechnen
df['price_per_room'] = df['price'] / df['total_rooms']

# 2. Altersgruppen des Gebäudes
df['building_age_group'] = pd.cut(df['building_age'], 
                                  bins=[-1, 5, 15, 30, 100], 
                                  labels=['Neubau (0-5)', 'Jung (6-15)', 'Mittel (16-30)', 'Alt (31+)'])

# 3. Luxus-Flag (sehr teure Wohnungen)
df['is_luxury'] = (df['price'] > 50_000_000).astype(int)

# 4. Größen-Kategorie
df['size_category'] = pd.cut(df['gross_sqm'], 
                             bins=[0, 80, 120, 200, 1000], 
                             labels=['Klein (<80m²)', 'Mittel (80-120)', 'Groß (121-200)', 'Sehr groß (>200)'])

print("Neue Features erstellt!")
print(df[['price', 'total_rooms', 'price_per_room', 'building_age_group', 'is_luxury', 'size_category']].head())
```

```python
# 5. Weitere sinnvolle Features

# Preis pro m² (bereits vorhanden, aber sicherstellen)
print("Preis pro m² existiert bereits:", 'price_per_sqm' in df.columns)

# Bezirk als Luxus- oder Normal-Bezirk kategorisieren (Beispiel)
luxury_districts = ['Beşiktaş', 'Şişli', 'Kadıköy', 'Bağdat Caddesi', 'Sarıyer', 'Beyoğlu']
df['district_category'] = df['district'].apply(
    lambda x: 'Luxus-Bezirk' if x in luxury_districts else 'Normal-Bezirk'
).astype('category')

# Stockwerk-Relation (auf welchem Stockwerk relativ zum Gebäude)
df['floor_ratio'] = df['floor'] / df['total_floors']

# Einfaches Furnished-Flag (binär)
df['is_furnished'] = df['furnished'].apply(lambda x: 1 if str(x).lower() in ['furnished', 'yes'] else 0)

print("Weitere Features erstellt!")
print(df[['district', 'district_category', 'floor_ratio', 'is_furnished']].head(8))
```

```python
# Finale Version mit allen Features speichern
df.to_csv('../01_data/istanbul_apartment_prices_final.csv', index=False)

print("Finales Dataset mit allen Features gespeichert!")
print("Datei: ../01_data/istanbul_apartment_prices_final.csv")
print(f"Shape: {df.shape} (Zeilen, Spalten)")
```

```python
# Kurze finale Übersicht
print("Verfügbare Spalten im finalen Dataset:")
print(df.columns.tolist())

print("\nBeispiel der neuen Features:")
print(df[['price', 'price_per_sqm', 'price_per_room', 'building_age_group', 
          'size_category', 'district_category', 'is_luxury']].head(5))
```

```python
import sqlite3
import os

# Verbindung zur Datenbank herstellen (wird automatisch erstellt)
db_path = '../01_data/istanbul_housing.db'
conn = sqlite3.connect(db_path)

print(f"SQLite Datenbank erfolgreich erstellt: {db_path}")
```

```python
# Finale Tabelle mit allen relevanten Spalten erstellen
create_table_sql = """
CREATE TABLE IF NOT EXISTS apartments (
    listing_id TEXT PRIMARY KEY,
    price INTEGER,
    price_per_sqm REAL,
    district TEXT,
    neighborhood TEXT,
    rooms INTEGER,
    halls INTEGER,
    total_rooms INTEGER,
    gross_sqm REAL,
    net_sqm REAL,
    floor REAL,
    floor_category TEXT,
    total_floors INTEGER,
    building_age INTEGER,
    building_type TEXT,
    building_condition TEXT,
    heating_type TEXT,
    fuel_type TEXT,
    bathroom_count REAL,
    furnished TEXT,
    usage_status TEXT,
    is_in_complex INTEGER,
    complex_name TEXT,
    orientation TEXT,
    maintenance_fee REAL,
    credit_eligible TEXT,
    deed_status TEXT,
    exchange TEXT,
    last_updated TEXT,
    scraped_at TEXT,
    
    -- Neue Features
    price_per_room REAL,
    building_age_group TEXT,
    is_luxury INTEGER,
    size_category TEXT,
    district_category TEXT,
    floor_ratio REAL,
    is_furnished INTEGER
);
"""

conn.execute(create_table_sql)
conn.commit()

print("Tabelle 'apartments' erfolgreich erstellt!")
```

```python
# Daten aus DataFrame in die SQLite-Tabelle laden
df.to_sql('apartments', conn, if_exists='replace', index=False)

print("Alle Daten erfolgreich in die Tabelle 'apartments' geladen!")
print(f"Anzahl Datensätze in der Datenbank: {len(df)}")
```

```python
# Top 10 teuerste Wohnungen
query1 = """
SELECT 
    district,
    neighborhood,
    price,
    gross_sqm,
    price_per_sqm,
    building_age,
    building_condition
FROM apartments 
ORDER BY price DESC 
LIMIT 10;
"""

top10 = pd.read_sql_query(query1, conn)
print("Top 10 teuerste Wohnungen:")
print(top10)
```

```python
# Durchschnittspreis pro Bezirk
query2 = """
SELECT 
    district,
    COUNT(*) as anzahl,
    ROUND(AVG(price), 0) as avg_price,
    ROUND(AVG(price_per_sqm), 0) as avg_price_per_sqm,
    ROUND(AVG(gross_sqm), 1) as avg_size
FROM apartments 
GROUP BY district 
ORDER BY avg_price DESC 
LIMIT 15;
"""

bezirk_preise = pd.read_sql_query(query2, conn)
print("Durchschnittspreise pro Bezirk:")
print(bezirk_preise)
```

```python
#Luxus vs Normal (mit neuen Features)
query3 = """
SELECT 
    district_category,
    COUNT(*) as anzahl,
    ROUND(AVG(price), 0) as avg_price,
    ROUND(AVG(price_per_sqm), 0) as avg_price_per_sqm,
    ROUND(AVG(building_age), 1) as avg_age
FROM apartments 
GROUP BY district_category;
"""

luxus = pd.read_sql_query(query3, conn)
print("Luxus-Bezirke vs Normal-Bezirke:")
print(luxus)
```

```python
#Teuerste / Attraktivste Bezirke für Expats (Preis pro m²)
query_expats1 = """
SELECT 
    district,
    COUNT(*) as anzahl_wohnungen,
    ROUND(AVG(price_per_sqm), 0) as avg_price_per_sqm,
    ROUND(AVG(price), 0) as avg_price,
    ROUND(AVG(gross_sqm), 1) as avg_size
FROM apartments 
GROUP BY district 
ORDER BY avg_price_per_sqm DESC 
LIMIT 12;
"""

expats1 = pd.read_sql_query(query_expats1, conn)
print("Top Bezirke nach Preis pro m² (besonders interessant für Expats):")
print(expats1)
```

```python
# Neubau vs Altbau in beliebten Bezirken
query_expats2 = """
SELECT 
    district,
    building_age_group,
    COUNT(*) as anzahl,
    ROUND(AVG(price_per_sqm), 0) as avg_price_per_sqm,
    ROUND(AVG(maintenance_fee), 0) as avg_maintenance
FROM apartments 
WHERE district IN ('Beşiktaş', 'Şişli', 'Kadıköy', 'Sarıyer', 'Beyoğlu', 'Maltepe', 'Bakırköy')
GROUP BY district, building_age_group
ORDER BY district, avg_price_per_sqm DESC;
"""

expats2 = pd.read_sql_query(query_expats2, conn)
print("Neubau vs Altbau in Expat-beliebten Bezirken:")
print(expats2)
```

```python
#Furnished + In Complex (sehr wichtig für Neuankömmlinge)
query_expats3 = """
SELECT 
    district,
    COUNT(*) as total,
    SUM(CASE WHEN is_furnished = 1 THEN 1 ELSE 0 END) as furnished,
    ROUND(100.0 * SUM(CASE WHEN is_furnished = 1 THEN 1 ELSE 0 END) / COUNT(*), 1) as percent_furnished,
    SUM(CASE WHEN is_in_complex = 1 THEN 1 ELSE 0 END) as in_complex
FROM apartments 
WHERE district IN ('Beşiktaş', 'Şişli', 'Kadıköy', 'Sarıyer', 'Beyoğlu')
GROUP BY district
ORDER BY percent_furnished DESC;
"""

expats3 = pd.read_sql_query(query_expats3, conn)
print("Furnished & In Complex – Ideal für Expats:")
print(expats3)
```

```python
# Durchschnittspreis für "komfortable" Wohnungen (3+ Zimmer, >100m²)
query_expats4 = """
SELECT 
    district,
    COUNT(*) as anzahl,
    ROUND(AVG(price), 0) as avg_price,
    ROUND(AVG(gross_sqm), 1) as avg_sqm
FROM apartments 
WHERE total_rooms >= 3 AND gross_sqm >= 100
GROUP BY district 
ORDER BY avg_price DESC 
LIMIT 10;
"""

expats4 = pd.read_sql_query(query_expats4, conn)
print("Familien- / Komfort-Wohnungen (3+ Zimmer, ≥100m²):")
print(expats4)
```

```python
# Verteilung der Wohnungsgrößen
query_small1 = """
SELECT 
    CASE 
        WHEN gross_sqm <= 60 THEN 'Single / Bis 60m²'
        WHEN gross_sqm <= 100 THEN 'Mittel (61-100m²)'
        ELSE 'Groß (>100m²)'
    END as size_group,
    COUNT(*) as anzahl,
    ROUND(100.0 * COUNT(*) / SUM(COUNT(*)) OVER(), 1) as prozent,
    ROUND(AVG(price), 0) as avg_price,
    ROUND(AVG(price_per_sqm), 0) as avg_price_per_sqm
FROM apartments 
GROUP BY size_group
ORDER BY anzahl DESC;
"""

size_dist = pd.read_sql_query(query_small1, conn)
print("Verteilung der Wohnungsgrößen:")
print(size_dist)
```

```python
# Small Apartments (≤ 60m²) – detailliert
query_small2 = """
SELECT 
    district,
    COUNT(*) as anzahl,
    ROUND(AVG(price), 0) as avg_price,
    ROUND(AVG(price_per_sqm), 0) as avg_price_per_sqm,
    ROUND(AVG(gross_sqm), 1) as avg_size,
    SUM(CASE WHEN is_in_complex = 1 THEN 1 ELSE 0 END) as in_complex,
    ROUND(100.0 * SUM(CASE WHEN is_in_complex = 1 THEN 1 ELSE 0 END) / COUNT(*), 1) as percent_in_complex
FROM apartments 
WHERE gross_sqm <= 60
GROUP BY district 
ORDER BY avg_price DESC 
LIMIT 12;
"""

small = pd.read_sql_query(query_small2, conn)
print("Small Apartments (bis 60m²) nach Bezirk:")
print(small)
```

```python
#Top 10 teuerste Single-Wohnungen (≤ 60m²)
query_small3 = """
SELECT 
    district,
    neighborhood,
    price,
    gross_sqm,
    price_per_sqm,
    building_age_group,
    is_in_complex,
    furnished
FROM apartments 
WHERE gross_sqm <= 60
ORDER BY price DESC 
LIMIT 10;
"""

top_small = pd.read_sql_query(query_small3, conn)
print("Top 10 teuerste Single-Wohnungen (≤ 60m²):")
print(top_small)
```

```python
# Asian Side vs European Side Vergleich
query_side = """
SELECT 
    CASE 
        WHEN district IN ('Kadıköy', 'Üsküdar', 'Maltepe', 'Ataşehir', 'Kartal', 
                         'Beykoz', 'Sancaktepe', 'Çekmeköy') 
        THEN 'Asian Side (Anatolian)'
        ELSE 'European Side (Thracian)'
    END as side,
    COUNT(*) as number_of_listings,
    ROUND(AVG(price), 0) as avg_price_try,
    ROUND(AVG(price_per_sqm), 0) as avg_price_per_sqm,
    ROUND(AVG(gross_sqm), 1) as avg_size_sqm,
    ROUND(AVG(building_age), 1) as avg_building_age
FROM apartments 
GROUP BY side;
"""

side_comparison = pd.read_sql_query(query_side, conn)
print("Asian Side vs European Side Comparison:")
print(side_comparison)
```

```python
# Small Apartments (≤ 60m²) nach Seite
query_small_side = """
SELECT 
    CASE 
        WHEN district IN ('Kadıköy', 'Üsküdar', 'Maltepe', 'Ataşehir', 'Kartal', 
                         'Beykoz', 'Sancaktepe', 'Çekmeköy') 
        THEN 'Asian Side (Anatolian)'
        ELSE 'European Side (Thracian)'
    END as side,
    COUNT(*) as number_of_small_apartments,
    ROUND(AVG(price), 0) as avg_price_try,
    ROUND(AVG(price_per_sqm), 0) as avg_price_per_sqm,
    ROUND(100.0 * AVG(CASE WHEN is_furnished = 1 THEN 1 ELSE 0 END), 1) as percent_furnished
FROM apartments 
WHERE gross_sqm <= 60
GROUP BY side;
"""

small_side = pd.read_sql_query(query_small_side, conn)
print("Small Apartments (≤ 60m²) - Asian vs European Side:")
print(small_side)
```
